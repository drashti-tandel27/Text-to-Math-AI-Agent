## Task 1: Text-to-Math Agent Overview

### 1. What is a Text-to-Math problem?

A Text-to-Math problem is a mathematical question written in simple natural language. The system understands the question and calculates the answer.

**Example:** A pen costs ₹10. What is the cost of 5 pens?

**Answer:** 10 × 5 = ₹50.

### 2. Why are agents useful for math reasoning?

Agents can understand a question, select a calculator tool, perform calculations, and explain the answer step by step. This helps reduce calculation mistakes.

### 3. Difference between a normal LLM response and agent-based reasoning

- **Normal LLM:** Generates an answer directly using its language model.
- **Agent-based reasoning:** Can use tools such as a calculator to perform calculations and then provide an answer with an explanation.

Agents are useful for mathematical problems because they can combine language understanding with external tools.
